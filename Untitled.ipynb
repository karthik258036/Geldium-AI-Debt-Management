```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)
```

```python
# Load the cleaned dataset

df = pd.read_csv("Delinquency_prediction_dataset_cleaned_median.csv")

# Display first 5 rows
df.head()
```

```python
# Check dataset shape
print("Dataset shape:", df.shape)

# Check column names
print("\nColumns:")
print(df.columns.tolist())

# Check data types and missing values
print("\nDataset information:")
df.info()
```

```python
# Check the target variable distribution

print(df["Delinquent_Account"].value_counts())

print("\nTarget percentages:")
print(df["Delinquent_Account"].value_counts(normalize=True) * 100)
```

```python
# Separate input features (X) and target variable (y)

X = df.drop(columns=["Customer_ID", "Delinquent_Account"])
y = df["Delinquent_Account"]

print("Features:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)

print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Identify numerical and categorical columns

numeric_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = X.select_dtypes(include=["str", "object"]).columns.tolist()

print("Numerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Preprocessing for numerical features
numeric_transformer = StandardScaler()

# Preprocessing for categorical features
categorical_transformer = OneHotEncoder(
    handle_unknown="ignore",
    drop="first"
)

# Combine both preprocessing methods
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully.")
```

```python
from sklearn.model_selection import train_test_split

# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set shape:", X_train.shape)
print("Testing set shape:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Create Logistic Regression model
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Create complete pipeline
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", logistic_model)
    ]
)

print("Logistic Regression model created successfully.")
```

```python
# Train the Logistic Regression model

model.fit(X_train, y_train)

print("Model training completed successfully.")
```

```python
# Predict class labels
y_pred = model.predict(X_test)

# Predict probabilities of delinquency
y_prob = model.predict_proba(X_test)[:, 1]

print("Predicted classes:")
print(y_pred[:10])

print("\nPredicted delinquency probabilities:")
print(y_prob[:10])
```

```python
from sklearn.metrics import classification_report

print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)
roc_auc = roc_auc_score(y_test, y_prob)

print("Model Evaluation Results")
print("------------------------")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
print(f"ROC-AUC   : {roc_auc:.4f}")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Non-Delinquent", "Delinquent"]
).plot()

plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import RocCurveDisplay

RocCurveDisplay.from_predictions(y_test, y_prob)

plt.title("Logistic Regression - ROC Curve")
plt.show()
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.10, 0.15, 0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50]

print("Threshold | Precision | Recall | F1")
print("--------------------------------------")

for threshold in thresholds:
    y_pred_threshold = (y_prob >= threshold).astype(int)

    precision = precision_score(
        y_test, y_pred_threshold, zero_division=0
    )

    recall = recall_score(
        y_test, y_pred_threshold, zero_division=0
    )

    f1 = f1_score(
        y_test, y_pred_threshold, zero_division=0
    )

    print(
        f"{threshold:8.2f} | "
        f"{precision:9.4f} | "
        f"{recall:6.4f} | "
        f"{f1:6.4f}"
    )
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_predict

# 5-fold stratified cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Get probability predictions using only training data
y_train_cv_prob = cross_val_predict(
    model,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba"
)[:, 1]

print("Cross-validation probability predictions generated.")
print("Number of predictions:", len(y_train_cv_prob))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = np.arange(0.05, 0.51, 0.01)

best_threshold = 0
best_f1 = 0

for threshold in thresholds:
    y_train_cv_pred = (
        y_train_cv_prob >= threshold
    ).astype(int)

    f1 = f1_score(
        y_train,
        y_train_cv_pred,
        zero_division=0
    )

    if f1 > best_f1:
        best_f1 = f1
        best_threshold = threshold

print("Best threshold:", round(best_threshold, 2))
print("Cross-validation F1:", round(best_f1, 4))
```

```python
# Apply the threshold selected from training cross-validation

y_pred_final = (
    y_prob >= best_threshold
).astype(int)

print("Final threshold:", round(best_threshold, 2))
print("Predicted classes:")
print(y_pred_final)
```

```python
final_accuracy = accuracy_score(y_test, y_pred_final)
final_precision = precision_score(
    y_test,
    y_pred_final,
    zero_division=0
)
final_recall = recall_score(
    y_test,
    y_pred_final,
    zero_division=0
)
final_f1 = f1_score(
    y_test,
    y_pred_final,
    zero_division=0
)
final_roc_auc = roc_auc_score(
    y_test,
    y_prob
)

print("Final Logistic Regression Results")
print("---------------------------------")
print(f"Threshold : {best_threshold:.2f}")
print(f"Accuracy  : {final_accuracy:.4f}")
print(f"Precision : {final_precision:.4f}")
print(f"Recall    : {final_recall:.4f}")
print(f"F1 Score  : {final_f1:.4f}")
print(f"ROC-AUC   : {final_roc_auc:.4f}")
```

```python
df["Employment_Status"] = df["Employment_Status"].replace({
    "EMP": "Employed",
    "employed": "Employed"
})

print(df["Employment_Status"].value_counts())
```

```python
# Recreate features and target after standardizing Employment_Status

X_final = df.drop(columns=["Customer_ID", "Delinquent_Account"])
y_final = df["Delinquent_Account"]

print("Feature shape:", X_final.shape)
print("Target shape:", y_final.shape)
```

```python
# Identify numerical and categorical features

numeric_features_final = X_final.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features_final = X_final.select_dtypes(
    include=["str", "object"]
).columns.tolist()

print("Numerical features:")
print(numeric_features_final)

print("\nCategorical features:")
print(categorical_features_final)
```

```python
# Final preprocessing pipeline

preprocessor_final = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features_final
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            ),
            categorical_features_final
        )
    ]
)

print("Final preprocessing pipeline created.")
```

```python
# Stratified train-test split

X_train_final, X_test_final, y_train_final, y_test_final = train_test_split(
    X_final,
    y_final,
    test_size=0.20,
    random_state=42,
    stratify=y_final
)

print("Training set:", X_train_final.shape)
print("Testing set:", X_test_final.shape)
```

```python
# Final Logistic Regression model

final_logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

final_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor_final),
        ("classifier", final_logistic_model)
    ]
)

# Train
final_model.fit(X_train_final, y_train_final)

print("Final Logistic Regression model trained successfully.")
```

```python
# Make predictions using the final model

y_pred_final_model = final_model.predict(X_test_final)

# Predict probability of delinquency
y_prob_final_model = final_model.predict_proba(X_test_final)[:, 1]

print("Predicted classes:")
print(y_pred_final_model[:20])

print("\nPredicted delinquency probabilities:")
print(y_prob_final_model[:20])
```

```python
# Calculate final evaluation metrics

final_accuracy = accuracy_score(
    y_test_final,
    y_pred_final_model
)

final_precision = precision_score(
    y_test_final,
    y_pred_final_model,
    zero_division=0
)

final_recall = recall_score(
    y_test_final,
    y_pred_final_model,
    zero_division=0
)

final_f1 = f1_score(
    y_test_final,
    y_pred_final_model,
    zero_division=0
)

final_roc_auc = roc_auc_score(
    y_test_final,
    y_prob_final_model
)

print("Final Logistic Regression Results")
print("---------------------------------")
print(f"Accuracy  : {final_accuracy:.4f}")
print(f"Precision : {final_precision:.4f}")
print(f"Recall    : {final_recall:.4f}")
print(f"F1 Score  : {final_f1:.4f}")
print(f"ROC-AUC   : {final_roc_auc:.4f}")
```
