```python
df_clean = df.copy()

df_clean["Employment_Status"] = df_clean["Employment_Status"].replace({
    "EMP": "Employed",
    "employed": "Employed",
    "retired": "Retired"
})

print("Employment status after standardization:")
print(df_clean["Employment_Status"].value_counts())
```

```python
print("Missing values after categorical cleaning:")
print(df_clean.isnull().sum())
```

```python
X = df_clean.drop(columns=["Customer_ID", "Delinquent_Account"])
y = df_clean["Delinquent_Account"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
numeric_features = [
    "Age",
    "Income",
    "Credit_Score",
    "Credit_Utilization",
    "Missed_Payments",
    "Loan_Balance",
    "Debt_to_Income_Ratio",
    "Account_Tenure"
]

categorical_features = [
    "Employment_Status",
    "Credit_Card_Type",
    "Location",
    "Month_1",
    "Month_2",
    "Month_3",
    "Month_4",
    "Month_5",
    "Month_6"
]

print("Numerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
```

```python
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        drop="first"
    ))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", logistic_model)
])

model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully!")
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)
roc_auc = roc_auc_score(y_test, y_prob)

print("Model Performance")
print("-" * 30)
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")
```

```python
print("Prediction distribution:")
print(pd.Series(y_pred).value_counts())

print("\nActual distribution:")
print(y_test.value_counts())

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
payment_columns = [
    "Month_1",
    "Month_2",
    "Month_3",
    "Month_4",
    "Month_5",
    "Month_6"
]

payment_mapping = {
    "On-time": 0,
    "Late": 1,
    "Missed": 2
}

payment_numeric = df_clean[payment_columns].replace(payment_mapping)

df_clean["Total_Missed"] = (payment_numeric == 2).sum(axis=1)
df_clean["Total_Late"] = (payment_numeric == 1).sum(axis=1)
df_clean["Payment_History_Score"] = payment_numeric.mean(axis=1)

df_clean[
    ["Total_Missed", "Total_Late", "Payment_History_Score", "Delinquent_Account"]
].groupby("Delinquent_Account").agg(
    ["mean", "median", "min", "max"]
)
```

```python
correlation_data = df_clean.select_dtypes(include=["int64", "float64"]).corr()

target_correlation = (
    correlation_data["Delinquent_Account"]
    .drop("Delinquent_Account")
    .sort_values(key=abs, ascending=False)
)

print("Correlation with Delinquent_Account:")
print(target_correlation)
```

```python
df_clean["High_Utilization"] = (
    df_clean["Credit_Utilization"] > 0.70
).astype(int)

df_clean["High_DTI"] = (
    df_clean["Debt_to_Income_Ratio"] > 0.40
).astype(int)

df_clean["Payment_Risk"] = (
    (df_clean["Total_Missed"] >= 3) |
    (df_clean["Total_Late"] >= 3)
).astype(int)

print(df_clean[
    ["High_Utilization", "High_DTI", "Payment_Risk"]
].value_counts())
```

```python
risk_features = [
    "High_Utilization",
    "High_DTI",
    "Payment_Risk"
]

for feature in risk_features:
    print("\n" + "=" * 50)
    print(feature)
    print("=" * 50)

    result = df_clean.groupby(feature)["Delinquent_Account"].agg(
        Count="count",
        Delinquent_Count="sum",
        Delinquency_Rate="mean"
    )

    result["Delinquency_Rate"] = result["Delinquency_Rate"] * 100

    print(result)
```

```python
from sklearn.ensemble import RandomForestClassifier

random_forest = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        max_depth=5,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42
    ))
])

random_forest.fit(X_train, y_train)

rf_pred = random_forest.predict(X_test)
rf_prob = random_forest.predict_proba(X_test)[:, 1]

print("Random Forest Performance")
print("-" * 35)
print(f"Accuracy : {accuracy_score(y_test, rf_pred):.4f}")
print(f"Precision: {precision_score(y_test, rf_pred, zero_division=0):.4f}")
print(f"Recall   : {recall_score(y_test, rf_pred, zero_division=0):.4f}")
print(f"F1 Score : {f1_score(y_test, rf_pred, zero_division=0):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, rf_prob):.4f}")
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

cv_results = cross_validate(
    model,
    X,
    y,
    cv=cv,
    scoring=scoring
)

print("5-Fold Cross-Validation Results")
print("-" * 40)

for metric in scoring.keys():
    scores = cv_results[f"test_{metric}"]
    print(
        f"{metric.upper():10s}: "
        f"{scores.mean():.4f} ± {scores.std():.4f}"
    )
```

```python
balanced_logistic = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

balanced_logistic.fit(X_train, y_train)

balanced_pred = balanced_logistic.predict(X_test)
balanced_prob = balanced_logistic.predict_proba(X_test)[:, 1]

print("Class-Weighted Logistic Regression")
print("-" * 45)
print(f"Accuracy : {accuracy_score(y_test, balanced_pred):.4f}")
print(f"Precision: {precision_score(y_test, balanced_pred, zero_division=0):.4f}")
print(f"Recall   : {recall_score(y_test, balanced_pred, zero_division=0):.4f}")
print(f"F1 Score : {f1_score(y_test, balanced_pred, zero_division=0):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, balanced_prob):.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, balanced_pred))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Confusion Matrix
ConfusionMatrixDisplay.from_predictions(
    y_test,
    balanced_pred,
    ax=axes[0],
    cmap="Blues"
)
axes[0].set_title("Class-Weighted Logistic Regression\nConfusion Matrix")

# ROC Curve
RocCurveDisplay.from_predictions(
    y_test,
    balanced_prob,
    ax=axes[1]
)
axes[1].plot([0, 1], [0, 1], linestyle="--")
axes[1].set_title("Class-Weighted Logistic Regression\nROC Curve")

plt.tight_layout()
plt.show()
```
